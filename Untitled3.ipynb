```python
class Graph:
    def __init__(self):
        self.graph = {}

    def add_edge(self, u, v):
        if u not in self.graph:
            self.graph[u] = []
        if v not in self.graph:
            self.graph[v] = []
        self.graph[u].append(v)
        # For an undirected graph, add the reverse edge as well
        # self.graph[v].append(u)

    def dfs_util(self, v, visited, result):
        visited.add(v)
        result.append(v)
        if v in self.graph:
            for neighbour in self.graph[v]:
                if neighbour not in visited:
                    self.dfs_util(neighbour, visited, result)

    def dfs(self, start_node):
        visited = set()
        result = []
        self.dfs_util(start_node, visited, result)
        return result

    def bfs(self, start_node):
        visited = set()
        queue = []
        result = []

        visited.add(start_node)
        queue.append(start_node)

        while queue:
            s = queue.pop(0)
            result.append(s)

            if s in self.graph:
                for neighbour in self.graph[s]:
                    if neighbour not in visited:
                        visited.add(neighbour)
                        queue.append(neighbour)
        return result

# Example Usage:
if __name__ == '__main__':
    g = Graph()
    g.add_edge('A', 'B')
    g.add_edge('A', 'C')
    g.add_edge('B', 'D')
    g.add_edge('B', 'E')
    g.add_edge('C', 'F')
    g.add_edge('E', 'F')

    print("Graph Adjacency List:", g.graph)

    print("\nDFS Traversal (starting from A):")
    dfs_path = g.dfs('A')
    print(dfs_path)

    print("\nBFS Traversal (starting from A):")
    bfs_path = g.bfs('A')
    print(bfs_path)

    # Example with a disconnected graph or different start node
    g2 = Graph()
    g2.add_edge(0, 1)
    g2.add_edge(0, 2)
    g2.add_edge(1, 2)
    g2.add_edge(2, 0)
    g2.add_edge(2, 3)
    g2.add_edge(3, 3)

    print("\nGraph 2 Adjacency List:", g2.graph)

    print("\nDFS Traversal (Graph 2, starting from 2):")
    dfs_path_g2 = g2.dfs(2)
    print(dfs_path_g2)

    print("\nBFS Traversal (Graph 2, starting from 2):")
    bfs_path_g2 = g2.bfs(2)
    print(bfs_path_g2)
```
